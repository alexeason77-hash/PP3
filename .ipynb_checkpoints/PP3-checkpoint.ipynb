```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def main():

#define data set name
    data_2024 = ('Alberta_power_usage_2024.csv')
    data_2021 = ('Alberta_power_usage_2021.csv')


    data_load = False
    end = False
    user_data_history = []
    regs = ['Calgary', 'Central', 'Edmonton', 'Northeast', 'Northwest', 'South']
    saved_results = {}
    while end != True:
        print(
            '====================================================\n'
            'ALBERTA POWER ANALYSIS TOOL\n'
            '====================================================\n'
            '1. Load Power Data sets\n'
            '2. View dataset summary\n'
            '3. Convert to daily / back to hourly\n'
            '4. Filter and explore (by region, month, date range)\n'
            '5. Perform calculations\n'
            '6. Generate visualizations\n'
            '7. Compare 2021 vs 2024\n'
            '8. Export results\n'
            '9. Exit'
        ) 
        
           
        
        user_select = int(input("Enter your selection: "))
        while user_select not in [1, 2, 3, 4, 5, 6, 7, 8, 9]:
            print('Invalid selection')
            user_select = int(input('Enter your selection: '))

#option 1
        if user_select == 1:
#user input year wanted
            data_year = int(input("what year of data(2021 or 2024)"))
            while data_year != 2021 and data_year != 2024:
                print("invalid input year")
                data_year = int(input("what year of data(2021 or 2024)"))

#tracks the history of the data loaded
            user_data_history.append(data_year)
#clean empty and duplacate spaces
            from PP3_Functions import data_clean
            if data_year == 2024:
                data_2024 = data_clean(data_2024)
            else:
                data_2021 = data_clean(data_2021)

#convert to data set to csv
            if data_year == 2024:
                data_2024.to_csv('clean_Alberta_power_usage_2024.csv',index=False)
            else:
                data_2021.to_csv('clean_Alberta_power_usage_2021.csv',index=False)

#assignes the data set of the year selected
            if data_year == 2021:
                dataset = data_2021
            elif data_year == 2024:
                dataset = data_2024
    
#adjust for 2024 leap_year
            if data_year == 2024:
                leap_choice = input('Do you want to include febuary 29th?(yes/no)')
                while leap_choice.lower() != 'yes' and leap_choice.lower() != 'no':
                    print('invalid choice please enter yes or no')
                    leap_choice = input('Do you want to include febuary 29th?(yes/no)')
                if leap_choice.lower() == 'no':    
                    from PP3_Functions import leap_del
                    dataset_leapdel = leap_del(dataset)
                    dataset = dataset_leapdel

#ensure the data was loaded
            data_load = True
#option 2
        if user_select == 2 and data_load == True:
            if data_year == 2021: 
                leap = 'no leap'
            else:
                if leap_choice == 'no':
                    leap = 'not included'
                else:
                    leap = 'included'
            if len(dataset) > 500:
                time_step = 'hours'
            else:
                time_step = 'days'     
            print(f"Year: {data_year} | leap day: {leap} | Time step: {time_step}")
            
            print()

            from PP3_Functions import region_stats
            print(f"{'Regions':<20}{'Mean':<10}{'Min':<10}{'Max':<10}")
            print('-'*50)
            rows = []
            for region in regs:
                average, minimum, maximum = region_stats(dataset, region)
                print(f"{region:<20}{average:<10.1f}{minimum:<10.1f}{maximum:<10.1f}")
                rows.append([region, average, minimum, maximum])
            
            summary = pd.DataFrame(rows, columns=['Region','Mean','Min','Max'])
            saved_results['summary']= summary
#option 3      
        if user_select == 3 and data_load == True:
#convert usage to by the day
            if len(dataset) > 500:
                from PP3_Functions import day_choice
                conv_choice = day_choice('day')
                    
                if conv_choice.lower() == 'yes':
                    from PP3_Functions import to_day
                    dataset = to_day(dataset)

            else:
                from PP3_Functions import day_choice
                conv_choice = day_choice('hour')
                
                if conv_choice.lower() == 'yes':
                    if data_year == 2024:
                        dataset = data_2024
                        if leap_choice == 'no':
                            dataset = dataset_leapdel
                    else:
                        dataset = data_2021
                    

#option 4
        if user_select == 4 and data_load == True:
            if len(dataset) > 500:
                datasetcopy = dataset.copy() 
                datasetcopy = to_day(datasetcopy)
            else:
                datasetcopy = dataset.copy()

            valid_days = datasetcopy['Days'].tolist()

            start = input('Enter the starting range date.(2024-01-01)')
            while start not in valid_days:
                print('Invalid starting point')
                start = input('Enter the starting range date.(2024-01-01)')

            end_date = input('Enter the ending range date.(2024-01-01)')
            while end_date not in valid_days or valid_days.index(end_date) < valid_days.index(start):
                print('Invalid ending point')
                end_date = input('Enter the ending range date.(2024-01-01)')  
            
                    
            starting_row = valid_days.index(start)
            ending_row = valid_days.index(end_date)
            restricted_data = datasetcopy.iloc[starting_row : ending_row + 1]

            selected_regions = []
            for r in range(6):
                selected_region = input('Enter your desired regions(enter end to stop): ')
                if selected_region.lower() == 'end':
                    break
                elif selected_region in regs:
                    if selected_region not in selected_regions:
                        selected_regions.append(selected_region)
                    else:
                        print('Region already selected')
                else:
                    print('Invalid region selection')
                
            filtered_data = restricted_data[['Days']+ selected_regions]
            
            
            
            
            
            




#option 7        
        if user_select == 7 and data_load == True:
            if 2024 not in user_data_history:
                data_2024 = data_clean(data_2024)
                data_2024.to_csv('clean_Alberta_power_usage_2024.csv',index=False)
            
            elif 2021 not in user_data_history: 
                data_2021 = data_clean(data_2021)
                data_2021.to_csv('clean_Alberta_power_usage_2021.csv',index=False)
        
        
        if user_select == 9:
            end = True
    



if __name__ == "__main__":
    main()
```
