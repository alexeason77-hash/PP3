```python
import numpy as np
import matplotlib.pyplot as plt

data_2024 = np.genfromtxt('Alberta_power_usage_2024.csv', delimiter = ',', skip_header = True)
data_2021 = np.genfromtxt('Alberta_power_usage_2021.csv', delimiter = ',', skip_header = True)


#clean the inputs



regions = {'Calgary': 1, 'Central': 2, 'Edmonton': 3, 'Northeast': 4, 'Northwest': 5, 'South': 6}

data_year = int(input("what year of data(2021 or 2024)"))

while data_year != 2021 and data_year != 2024:
    print("invalid input year")
    data_year = int(input("what year of data(2021 or 2024)"))

if data_year == 2021:
    dataset = data_2021
elif data_year == 2024:
    dataset = data_2024

#gets rid of febuary 29 in data that has it 
def leap_del(dataset)
    if len(dataset) > 8770:
        datascale = dataset.copy()
        for i in range(1440, 1416, -1):
            datascale = np.delete(datascale, i, axis=0)
    #print(datascale)
print()




user_region = input("enter the region")

region_check = False
while region_check == False:
    for key in regions:
        region_check = user_region == key
        if region_check == True:
            break
    if region_check == False:
        print('invalid region entry')
        user_region = input("enter the region")

        
    
    

region_num = regions[user_region]


from PP3_Functions import power_h_d

power_per_day = power_h_d(dataset, region_num)    
print(power_per_day)

output_data = open("output_file.csv", "w")
output_data.write(f"Day({user_region}), Total Power({user_region})\n")
for day, total in power_per_day:
    output_data.write(f"{day}, {total}\n")
output_data.close()
```
