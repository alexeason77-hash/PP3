```python
import pandas as pd

data_2024 = ('Alberta_power_usage_2024.csv')
data_2021 = ('Alberta_power_usage_2021.csv')



def data_clean(data_file):

    dataset = pd.read_csv(data_file)

    print(dataset.info())

    print("empty")
    print(dataset.isnull().sum())
    
    dataset.drop_duplicates(inplace=True)
    dataset.fillna(dataset.mean(numeric_only=True), inplace = True)
    return(dataset)

clean_2024 = data_clean(data_2024)
clean_2021 = data_clean(data_2021)




```
